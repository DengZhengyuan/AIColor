## Plot all data

```python
lists_nozero = [
    col[col != 0].values
    for _, col in df.items()
]

lists_norm = [
    arr / np.max(arr)
    for arr in lists_nozero
    if len(arr) > 0
]
```

```python
import matplotlib.pyplot as plt
import math

group_size = 25
n_groups = math.ceil(len(lists_norm) / group_size)

for i in range(n_groups):
    start = i * group_size
    end   = min((i + 1) * group_size, len(lists_norm)-4)

    plt.figure(figsize=(8, 2))
    parts = plt.violinplot(
        lists_norm[start:end],
        showmeans=True,
        showextrema=True
    )

    for pc in parts["bodies"]:
        pc.set_alpha(0.6)

    plt.xticks([])
    plt.ylabel("Normalized value")
    plt.tight_layout()
    plt.show()
```

```python
lists_norm_new = df.copy()

last_cols = lists_norm_new.columns[-4:]

for col in last_cols:
    x = lists_norm_new[col].values
    xmin = x.min()
    xmax = x.max()

    if xmax > xmin:
        lists_norm_new[col] = (x - xmin) / (xmax - xmin)
    else:
        lists_norm_new[col] = 0.0
```

```python
data = [lists_norm_new[col].values for col in last_cols]

plt.figure(figsize=(6, 2))

parts = plt.violinplot(
    data,
    showmeans=True,
    showextrema=True
)

# 调整透明度
for pc in parts["bodies"]:
    pc.set_alpha(0.6)

plt.xticks(range(1, len(last_cols) + 1), last_cols, rotation=0)
plt.ylabel("Normalized value (0–1)")
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler, RobustScaler, QuantileTransformer

# 取最后 4 列（如果你不是“最后4列”，把这里改成你想要的列名列表即可）
last_cols = df.columns[-4:]
X = df[last_cols].to_numpy(dtype=float)  # shape: (n_samples, 4)
n_samples = X.shape[0]

# --- 4 种归一化方法 ---
# 1) Min-Max -> [0,1]
mm = MinMaxScaler()
X_minmax = mm.fit_transform(X)

# 2) Robust (median/IQR) -> 通常不在[0,1]，但抗 outlier
rs = RobustScaler()
X_robust = rs.fit_transform(X)

# 3) Quantile -> 映射到均匀分布(0~1)，最“铺开”
qt = QuantileTransformer(
    output_distribution="uniform",
    n_quantiles=min(1000, n_samples),
    random_state=0
)
X_quantile = qt.fit_transform(X)

# 4) log1p（若有负数则整体平移到 >=0），再 Min-Max 到[0,1] 方便比较
X_shift = X.copy()
min_all = np.nanmin(X_shift)
if min_all < 0:
    X_shift = X_shift - min_all  # 平移使最小值为0

X_log = np.log1p(X_shift)
X_log_minmax = MinMaxScaler().fit_transform(X_log)

methods = [
    ("Min-Max (0–1)", X_minmax),
    ("Robust (median/IQR)", X_robust),
    ("Quantile -> Uniform (0–1)", X_quantile),
    ("log1p + Min-Max (0–1)", X_log_minmax),
]

# --- 画图：每种方法一张图，图里 4 个 violin ---
for title, Xn in methods:
    data = [Xn[:, j] for j in range(Xn.shape[1])]

    plt.figure(figsize=(7, 2.6))
    parts = plt.violinplot(data, showmeans=True, showextrema=True)

    for pc in parts["bodies"]:
        pc.set_alpha(0.6)

    plt.xticks(range(1, len(last_cols) + 1), last_cols)
    plt.ylabel("Value")
    plt.title(title)
    plt.tight_layout()
    plt.show()
```
